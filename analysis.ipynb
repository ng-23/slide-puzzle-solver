# Analysis

```python
from dotenv import load_dotenv
import os

load_dotenv()
```

## Algorithm Performance - BFS

```python
import pandas as pd

bfs_data = pd.read_csv(os.getenv('BFS_DATA_PATH'))

print(f'BFS performance data:\n{bfs_data.head(n=15)}')
print(f'BFS average time to solve: {bfs_data['solve_time'].mean()} seconds')
print(f'BFS average number of moves to solve: {bfs_data['num_moves'].mean()}')
print(f'BFS median number of moves to solve: {bfs_data['num_moves'].median()}')
print(f'BFS summary statistics:\n{bfs_data.describe()}')
```
