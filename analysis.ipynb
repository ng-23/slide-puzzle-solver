## Search Space Visualization

```python
from SlidePuzzle import SlidePuzzle
from pyvis.network import Network
import tkinter as tk

root = tk.Tk()
solve_algo = 'pc_bfs'
n_nodes = 20
solve_config = {
    'n_nodes': n_nodes,
}
seed = 1

puzzle = SlidePuzzle(
    root,
    solve_algo=solve_algo,
    solve_config=solve_config,
    seed=seed,
)

graph = puzzle.precompute_search_space(puzzle.current_state, puzzle.empty_pos, puzzle.goal_state, n_nodes=n_nodes)

net = Network(
    notebook=False,
    directed=True,
)

for i, node in enumerate(graph.keys()):
    s = ''
    for r in node:
        s = s + f'{' '*3}'.join(map(str, r)) + '\n'
    net.add_node(str(node), label=f'State{i}\n{s}', size=25)

for i, (node, neighbors) in enumerate(graph.items()):
    for neighbor in neighbors:
        if neighbor in graph and neighbor != node:
            move = graph[node][neighbor]
            net.add_edge(str(node), str(neighbor), title=f'{list(move)}')

net.toggle_physics(False)

net.show('search_space.html', notebook=False)
```
